```python
import sys
import platform
from datetime import datetime
from importlib.metadata import version, PackageNotFoundError

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "ldtk": "ldtk",
    "batman": "batman-package",
    "scipy": "scipy",
    "emcee": "emcee",
    "corner": "corner",
}

lines = [
    f"Recorded: {datetime.now().astimezone().isoformat(timespec='seconds')}",
    f"Python: {sys.version.replace(chr(10), ' ')}",
    f"Executable: {sys.executable}",
    f"Platform: {platform.platform()}",
    "",
]

for name, distribution in packages.items():
    try:
        lines.append(f"{distribution}=={version(distribution)}")
    except PackageNotFoundError:
        lines.append(f"{name}: version metadata not found")

text = "\n".join(lines)
print(text)

with open("package_versions.txt", "w", encoding="utf-8") as f:
    f.write(text + "\n")
```

```python
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "pip", "freeze"],
    capture_output=True, text=True, check=True
)

with open("requirements.txt", "w", encoding="utf-8") as f:
    f.write(result.stdout)
```
